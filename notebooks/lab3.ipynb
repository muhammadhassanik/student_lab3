# Part 6: Indexing and slicing






```python
# Setup: The scores array (3 students x 4 subjects)
scores = np.array([[45, 78, 90, 60],
                   [32, 65, 70, 55],
                   [88, 92, 79, 95]])

# 1. Select the first two subjects of the last student
print("1. Last student, first two subjects:", scores[-1, :2])

# 2. Select the third subject for students 1 and 2 only
print("2. Students 1 & 2, third subject:", scores[0:2, 2])

# 3. Select the 2x2 block at the top right
print("3. Top right 2x2 block:\n", scores[0:2, 2:4])

# 4. Show that editing a slice changes the original, but .copy() does not
slice_view = scores[0, :2]
slice_view[0] = 0 # This changes the original array
print("\n4a. Original array after slice edit (top left is now 0):\n", scores)

copy_view = scores[1, :2].copy()
copy_view[0] = 999 # This does NOT change the original array
print("\n4b. Original array after copy edit (remains unchanged):\n", scores)
```

# Part 7: Reshaping arrays

```python
# 1. Create a 1D array of 12 elements and reshape it to a 3x4 grid
arr = np.arange(12)
reshaped_arr = arr.reshape(3, 4)
print("1. Reshaped to 3x4:\n", reshaped_arr)

# 2. Transpose the 2D array (swap rows and columns)
transposed_arr = reshaped_arr.T
print("\n2. Transposed array:\n", transposed_arr)

# 3. Flatten the 2D array back into a 1D format
flattened_arr = reshaped_arr.flatten()
print("\n3. Flattened array:\n", flattened_arr)
```

## Part 10: Element-wise, matrix operations and linear algebra

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[10, 20], [30, 40]])
print(A * B)
print(A @ B)
print(A.T)

X = np.array([[1, 2, 3], [4, 5, 6]])
print((X @ X.T).shape)

M = np.array([[2, 1], [1, 3]])
rhs = np.array([5, 10])
sol = np.linalg.solve(M, rhs)
print(sol, np.linalg.det(M), np.allclose(M @ sol, rhs))
print(np.linalg.inv(M))
print(np.linalg.norm(np.array([3, 4])))
```

```python
# 1. Predict and test matrix shapes
print("Shape of (4, 3) @ (3, 5):", (np.ones((4, 3)) @ np.ones((3, 5))).shape)
try:
    np.ones((4, 3)) @ np.ones((4, 3))
except ValueError as e:
    print("Error for (4, 3) @ (4, 3):", e)

# 2. Weighted average
marks3 = np.array([[80, 70, 90], [60, 50, 40]])
weights = np.array([0.5, 0.3, 0.2])
weighted_marks = marks3 @ weights
print("\nWeighted marks:", weighted_marks)

# 3. Solve equations 3x+2y=12 and x-y=1
M_eq = np.array([[3, 2], [1, -1]])
rhs_eq = np.array([12, 1])
sol_eq = np.linalg.solve(M_eq, rhs_eq)
print("\nSolution for x and y:", sol_eq)
print("Verify:", np.allclose(M_eq @ sol_eq, rhs_eq))

# 4. Compute np.dot
dot_product = np.dot([1, 2, 3], [4, 5, 6])
print("\nDot product:", dot_product)

```
