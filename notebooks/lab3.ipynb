# Part 6: Indexing and slicing






```python
# Setup: The scores array (3 students x 4 subjects)
scores = np.array([[45, 78, 90, 60],
                   [32, 65, 70, 55],
                   [88, 92, 79, 95]])

# 1. Select the first two subjects of the last student
print("1. Last student, first two subjects:", scores[-1, :2])

# 2. Select the third subject for students 1 and 2 only
print("2. Students 1 & 2, third subject:", scores[0:2, 2])

# 3. Select the 2x2 block at the top right
print("3. Top right 2x2 block:\n", scores[0:2, 2:4])

# 4. Show that editing a slice changes the original, but .copy() does not
slice_view = scores[0, :2]
slice_view[0] = 0 # This changes the original array
print("\n4a. Original array after slice edit (top left is now 0):\n", scores)

copy_view = scores[1, :2].copy()
copy_view[0] = 999 # This does NOT change the original array
print("\n4b. Original array after copy edit (remains unchanged):\n", scores)
```

## Part 10: Element-wise, matrix operations and linear algebra

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[10, 20], [30, 40]])
print(A * B)
print(A @ B)
print(A.T)

X = np.array([[1, 2, 3], [4, 5, 6]])
print((X @ X.T).shape)

M = np.array([[2, 1], [1, 3]])
rhs = np.array([5, 10])
sol = np.linalg.solve(M, rhs)
print(sol, np.linalg.det(M), np.allclose(M @ sol, rhs))
print(np.linalg.inv(M))
print(np.linalg.norm(np.array([3, 4])))
```

## Part 12: Performance measurement: loop vs vectorized

```python
import time, timeit, tracemalloc
N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2

# Tool 1: perf_counter
for label, func, data in (('loop', loop_double, py_list), ('numpy', numpy_double, np_arr)):
    start = time.perf_counter()
    func(data)
    print(label, round(time.perf_counter() - start, 5), 'seconds')

# Tool 2: timeit 
n = 20
t_loop = timeit.timeit('loop_double(py_list)', globals=globals(), number=n) / n
t_np = timeit.timeit('numpy_double(np_arr)', globals=globals(), number=n) / n
print('loop', round(t_loop, 5), 'numpy', round(t_np, 5), 'ratio', round(t_loop / t_np, 1))

# Tool 3: tracemalloc
for label, make in (('list', lambda: [x * 2 for x in range(N)]), ('array', lambda: np.arange(N) * 2)):
    tracemalloc.start()
    obj = make()
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    print(label, 'current MB', round(current / 1e6, 1), 'peak MB', round(peak / 1e6, 1))
    
```

```python
# 1. Size sweep
rows = []
print("N\t\tloop_s\tnumpy_s\tratio")
for N_test in [1_000, 100_000, 1_000_000]:
    n_test = 100 if N_test < 1_000_000 else 20
    py_list_test = list(range(N_test))
    np_arr_test = np.arange(N_test)
    t_loop_test = timeit.timeit('loop_double(py_list_test)', globals=globals(), number=n_test) / n_test
    t_np_test = timeit.timeit('numpy_double(np_arr_test)', globals=globals(), number=n_test) / n_test
    ratio = t_loop_test / t_np_test
    rows.append([N_test, round(t_loop_test, 5), round(t_np_test, 5), round(ratio, 1)])
    print(f"{N_test}\t{t_loop_test:.5f}\t{t_np_test:.5f}\t{ratio:.1f}")

print("\nConclusion: The ratio grows significantly with N, showing vectorization scales much better for large arrays.")

# 2. Better timing
best_loop = min(timeit.repeat('loop_double(py_list)', globals=globals(), number=20, repeat=5)) / 20
best_np = min(timeit.repeat('numpy_double(np_arr)', globals=globals(), number=20, repeat=5)) / 20
print(f"\nBest loop: {best_loop:.5f}s, Best numpy: {best_np:.5f}s")
print("The best of several runs is fairer because it ignores runs slowed down by background OS interference.")
```
